```python
import numpy as np 
import pandas as pd
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler 
```

```python
df=pd.read_csv("train_df_1.csv")
df.drop(columns="ID",axis=1,inplace=True )
df.head(5)
df.info()
```

```python
df["Datetime"] = pd.to_datetime(
    df["Data"] + " " + df["Time"],
    format="%d-%m-%Y %H:%M:%S"
)
```

```python
df.head()
```

```python
df.info()
```

```python
df.head(5)
```

```python
df["hour"] = df["Datetime"].dt.hour
df["month"] = df["Datetime"].dt.month
df["day"] = df["Datetime"].dt.day
df["weekday"] = df["Datetime"].dt.weekday
df["year"] = df["Datetime"].dt.year
df.head(5)
```

```python
df.drop(columns=["Time","Datetime"],axis=1,inplace=True )
```

```python
df.drop(columns="Data",axis=1,inplace=True )
```

```python
df["TimeSunRise"] = pd.to_datetime(df["TimeSunRise"])
df["TimeSunSet"] = pd.to_datetime(df["TimeSunSet"])

df["sunrise_hour"] = df["TimeSunRise"].dt.hour
df["sunset_hour"] = df["TimeSunSet"].dt.hour
df.drop(["TimeSunRise", "TimeSunSet"], axis=1, inplace=True)
```

```python
df.drop("UNIXTime", axis=1, inplace=True)
df.head(5)
```

```python
df.drop(columns="year", axis=1, inplace=True)
df.isnull().sum()
```

```python
df.fillna(df.median(), inplace=True)
```

```python
df.head(5)
```

```python
df.to_csv("you_dataset.csv",index=False)
```
