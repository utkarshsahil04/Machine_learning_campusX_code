```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,accuracy_score,mean_squared_error
import numpy as np 
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X.shape
print(X_train.shape,y_train.shape)
```

```python
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
reg.coef_,reg.intercept_
```

```python
y_pred=reg.predict(X_test)
lr_r2_score=r2_score(y_test,y_pred)
```

```python
print(mean_squared_error(y_test,y_pred))
print(np.mean((y_test-y_pred)**2))
```

```python
class BGDRegressor:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.coef_=None
        self.intercept_=None
        self.lr=learning_rate
        self.epochs=epochs

    def fit(self,X_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(X_train.shape[1])
        for i in range(self.epochs):
            y_hat=self.intercept_ + np.dot(X_train,self.coef_)
            cost=np.mean((y_hat-y_train)**2)
            print(f"Epoch:{i} , cost:{cost}")
            intercept_der=-2*np.mean(y_train-y_hat)
            self.intercept_=self.intercept_-self.lr*intercept_der

            coeff_der=-2*np.dot((y_train-y_hat),X_train)/X_train.shape[0]
            self.coef_=self.coef_-self.lr*coeff_der
        print(f"Intercept {self.intercept_}\n coef_: {self.coef_}")
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
bgd=BGDRegressor(learning_rate=0.15,epochs=50000)
```

```python
bgd.fit(X_train,y_train)
```

```python
Y_batch_pred=bgd.predict(X_test)
```

```python
bgd_r2_score=r2_score(y_test,Y_batch_pred)
bgd_r2_score
```

```python

# BGD
bgd_train_mse = mean_squared_error(
    y_train,
    bgd.predict(X_train)
)

bgd_test_mse = mean_squared_error(
    y_test,
    bgd.predict(X_test)
)

# Linear Regression
lr_train_mse = mean_squared_error(
    y_train,
    reg.predict(X_train)
)

lr_test_mse = mean_squared_error(
    y_test,
    reg.predict(X_test)
)

print("BGD Train:", bgd_train_mse)
print("BGD Test :", bgd_test_mse)

print("LR Train :", lr_train_mse)
print("LR Test  :", lr_test_mse)
```
