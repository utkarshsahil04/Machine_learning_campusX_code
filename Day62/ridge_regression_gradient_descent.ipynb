```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score
import numpy as np 
from  sklearn.datasets import load_diabetes
```

```python
X,Y=load_diabetes(return_X_y=True)
print(f"X {X.shape}, y: {Y.shape}")
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print(X_train.shape,y_train.shape)
```

```python
reg=SGDRegressor(penalty="l2",max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
reg.fit(X_train,y_train)
y_pred=reg.predict(X_test)
print(f"r2_score {r2_score(y_test,y_pred)}")
```

```python
from sklearn.linear_model import Ridge

reg = Ridge(alpha=0.001, max_iter=500,solver='sparse_cg')
```

```python
reg.fit(X_train,y_train)

y_pred = reg.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(reg.coef_)
# print(reg.intercept_)
```

```python
class ourridge:
    def __init__(self,epochs,learning_rate,alpha):
        self.epochs=epochs
        self.learning_rate=learning_rate
        self.alpha=alpha
        self.coef_=None
        self.intercept_=None
    def fit(self,X_train,y_train):
        self.coef_=np.ones(X_train.shape[1])
        self.intercept_=0
        weight=np.insert(self.coef_,0,self.intercept_)
        X_train=np.insert(X_train,0,1,axis=1)

        for i in range (self.epochs):
            weight_der=np.dot(X_train.T,X_train).dot(weight)-np.dot(y_train.T,X_train)+self.alpha*weight
            print(f"weight derivative {i}:{weight_der}")
            weight=weight-self.learning_rate*weight_der
            print(f"weight  :{weight}")
        self.coef_=weight[1:]
        self.intercept_=weight[0]
    def predict(self,y_train):
        return np.dot(y_train,self.coef_) + self.intercept_
```

```python
reg=ourridge(epochs=200,learning_rate=0.005,alpha=0.001)
reg.fit(X_train,y_train)
y_pred1=reg.predict(X_test)
print(y_pred1)
```
